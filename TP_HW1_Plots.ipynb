```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter
rho_0 = 1.0  # Anfangsdichte
alpha = 1.0  # Beispielwert für alpha
beta = 2.0   # Beispielwert für beta
t = np.linspace(0, 2, 100)  # Zeit von 0 bis 3 in 100 Schritten

# Dichteprofilgleichung
rho = rho_0 * np.exp((alpha + beta) * t)

# Normalize rho
rho_normalized = rho / np.max(rho)

# Plot
plt.figure(figsize=(10, 6))
plt.plot(t, rho_normalized, label=r'$\rho(t) = \rho_0 e^{(\alpha + \beta)t}$ (normalized)')
plt.xlabel('Time (s)')
plt.ylabel('Normalized Density (-)')
plt.yticks(np.arange(0, 1.2, step=0.2))
plt.xticks(np.arange(0, 2.2, step=0.5))
plt.title('Normalized Density Profile of the Air in the Bottle Over Time')
plt.legend()
plt.grid(True)
plt.savefig('density_profile_Homework1.png')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter
rho_0 = 1.225  # Anfangsdichte der Luft in kg/m³ (Standarddichte auf Meereshöhe)
alpha = 0.010  # Beispielwert für alpha
beta = 0.020   # Beispielwert für beta
t = np.linspace(0, 20, 100)  # Zeit von 0 bis 20 Sekunden in 100 Schritten

# Dichteprofilgleichung in kg/m³
rho = rho_0 * np.exp((alpha + beta) * t)  # Dichte in kg/m³

# Plot
plt.figure(figsize=(10, 6))
plt.plot(t, rho, label=r'$\rho(t) = \rho_0 e^{(\alpha + \beta)t}$')
plt.xlabel('Time (s)', fontsize=12)
plt.ylabel('Density (kg/m³)', fontsize=12)
plt.title('Density Profile of the Air in the Bottle Over Time', fontsize=14)
plt.legend(fontsize=10)
plt.grid(True)

# Speichern und anzeigen
plt.savefig('density_profile_physical_units_density.png')
plt.show()
```
